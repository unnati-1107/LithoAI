```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

cm = np.array([[242, 51],
               [13, 8]])

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Pass", "Fail"]
)

disp.plot(cmap="Blues", values_format="d")
plt.title("LithoAI - Final Test Confusion Matrix")
plt.savefig("LithoAI_Confusion_Matrix.png",
            dpi=300, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ["Pass", "Fail"]
counts = [1463, 104]

plt.figure(figsize=(7, 5))
plt.bar(labels, counts)
plt.title("SECOM Dataset - Class Distribution")
plt.xlabel("Manufacturing Outcome")
plt.ylabel("Number of Observations")

for i, value in enumerate(counts):
    plt.text(i, value + 20, str(value), ha="center")

plt.savefig("LithoAI_Class_Distribution.png",
            dpi=300, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

steps = [
    "SECOM Dataset\n(1,567 observations, 590 features)",
    "Stratified Train-Test Split",
    "Training Data Split\n(Sub-training + Validation)",
    "Missing-Value Feature Filtering\n(590 → 566 features)",
    "Mean Imputation",
    "Constant Feature Removal\n(566 → 444 features)",
    "Feature Standardization",
    "SMOTE on Sub-training Data Only",
    "Logistic Regression Training",
    "Validation-Based Threshold Selection\n(Selected threshold: 0.1)",
    "Final Evaluation on Independent Test Set"
]

fig, ax = plt.subplots(figsize=(12, 17))
ax.set_xlim(0, 10)
ax.set_ylim(0, len(steps) * 1.5)
ax.axis("off")

for i, step in enumerate(steps):
    y = len(steps) * 1.5 - (i + 1) * 1.5 + 0.2

    box = FancyBboxPatch(
        (1.5, y), 7, 0.9,
        boxstyle="round,pad=0.08",
        edgecolor="black",
        facecolor="lightblue"
    )

    ax.add_patch(box)
    ax.text(5, y + 0.45, step,
            ha="center", va="center", fontsize=10)

    if i < len(steps) - 1:
        ax.annotate(
            "",
            xy=(5, y - 0.45),
            xytext=(5, y),
            arrowprops=dict(arrowstyle="->", lw=1.5)
        )

plt.title("LithoAI - Experimental Workflow", fontsize=16)
plt.savefig("LithoAI_Workflow.png",
            dpi=300, bbox_inches="tight")
plt.show()
```
